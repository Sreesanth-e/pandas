# Data cleaning
 
Treat the Nulls
Treat the duplicates
Populate missing rows
Drop unneeded columns
Split columns


```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('netflix1.csv')
```

```python
df.head()
```

```python
df.columns
```

```python
df.drop(columns= 'rating', inplace= True)
```

```python
df.columns
```

```python
new_column_name = []
for i in df.columns:
    new_column_name.append(i.capitalize())
```

```python
df.columns = new_column_name
```

```python
df.columns
```

```python
df.duplicated().sum()
```

```python
df["Show_id"] = df["Show_id"].apply(lambda x: x.split("s")[1])
```

```python
df.head()
```

```python
df["Show_id"] = df["Show_id"].astype(int)
```

```python
type(df["Show_id"][1])
```

```python
df["Date_added"] = df["Date_added"].apply(lambda x: x.replace("/","-"))
```

```python
df.head()
```

```python
plt.scatter(df['Country'],df['Type'])
plt.title("Scatter plot")
plt.xlabel("Country")
plt.ylabel("Type")
plt.show()
```
