```python
import pandas as pd
```

```python
df = pd.read_excel('Customer Call List.xlsx')
print(df.columns)
df .head()
df.duplicated().value_counts()
```

```python
print(df.duplicated().value_counts())
df.duplicated().sum()
df.drop_duplicates(inplace=True)
df.duplicated().sum()
```

```python
df = df.drop(columns= ['Not_Useful_Column'])
df.head()
```

```python
df['Last_Name'].isna()
```

```python
# df['Last_Name'] = df['Last_Name'].str.lstrip('...')
# df['Last_Name'] = df['Last_Name'].str.lstrip('/')
# df['Last_Name'] = df['Last_Name'].str.rstrip('_')
df['Last_Name'] = df['Last_Name'].str.strip('123./_')
df['Last_Name']
```

```python
df['Phone_Number']= df['Phone_Number'].str.replace('[^a-zA-Z0-9]' ,'',regex=True)

df['Phone_Number'] = df['Phone_Number'].apply(lambda x: str(x))
df['Phone_Number'] = df['Phone_Number'].apply(lambda x: x[0:3] + '-'+ x[3:6] + '-' + x[6:10])
df['Phone_Number'] = df['Phone_Number'].str.replace('nan--', '')
df['Phone_Number'] = df['Phone_Number'].str.replace('Na--', '')
df
```

```python
df[['Street address' ,'State','Zip code']]= df['Address'].str.split(',',n=2, expand = True)
df
```

```python
df['Paying Customer'] = df['Paying Customer'].replace('Yes','Y')
df['Paying Customer'] = df['Paying Customer'].replace('No','N')
print(df['Paying Customer'])
df['Do_Not_Contact'] = df['Do_Not_Contact'].replace('Yes','Y')
df['Do_Not_Contact'] = df['Do_Not_Contact'].replace('No','N')
df['Do_Not_Contact']
print(df)
```

```python
df = df.replace('N/a' ,'')
df = df.fillna(" ")
df
```

```python
for x in df.index:
    if df.loc[x,'Do_Not_Contact'] == 'Y' :
        df.drop(x , inplace=True)

df
```

```python
for x in df.index:
    if df.loc[x,'Phone_Number'] == '' :
        df.drop(x , inplace=True)

df
```

```python
df = df.reset_index(drop= True)
print(df.info())
df.describe(include = 'all')
```
