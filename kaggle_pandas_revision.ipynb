Summary Functions and Maps

```python
print(data.SepalLengthCm.describe)
```

```python
print(data.SepalLengthCm.mean())
```

```python
print(data.SepalLengthCm.unique())
```

```python
mean = data.SepalLengthCm.mean()
data.SepalLengthCm.map(lambda p: p - mean)
```

```python
mean_data = data.SepalLengthCm.mean()
def mean(data):
    data.SepalLengthCm = data.SepalLengthCm - mean_data
    return data

print(data.apply(mean  , axis='columns'))
```

Grouping and Sorting

```python
data.groupby('SepalLengthCm').PetalLengthCm.count()
```

```python
data.groupby('SepalLengthCm').PetalLengthCm.min()
```

```python
data.groupby('SepalLengthCm').apply( lambda p : p.PetalLengthCm.iloc[0])
```

```python
countries_reviewed  = data.groupby(['SepalLengthCm','SepalWidthCm']).PetalLengthCm.agg([len,max,min])
countries_reviewed
```

```python
data.sort_values(by='SepalLengthCm', ascending= False)
```
