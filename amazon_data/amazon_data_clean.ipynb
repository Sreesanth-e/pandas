4 . Discounted price
Discounted_Price (INR) : Remove ₹, Rs., commas and spaces, convert data type to Float, all values should be > 0.

```python
df
```

```python
df['Discounted_Price (INR)'].isnull().sum()
df.loc[~df['Discounted_Price (INR)'].str.contains('₹|Rs|,|\s', na=False), ['Discounted_Price (INR)']]

```

```python
# df['Discounted_Price (INR)'] = df['Discounted_Price (INR)'].str.replace('₹' , '')
# df['Discounted_Price (INR)'] = df['Discounted_Price (INR)'].str.replace('Rs. ' , '')
# df['Discounted_Price (INR)'] = df['Discounted_Price (INR)'].str.replace(',' , '')
df['Discounted_Price (INR)'] = df['Discounted_Price (INR)'].str.replace('₹|Rs\. |,', '' , regex=True)
df['Discounted_Price (INR)']
```

```python
df['Discounted_Price (INR)'] = pd.to_numeric(df['Discounted_Price (INR)'].str.strip().str.replace('\s+', '', regex=True),errors='coerce')
df['Discounted_Price (INR)'] = df['Discounted_Price (INR)'].astype('float')

```

```python
df['Discounted_Price (INR)'].dtype
```

```python
df[df['Discounted_Price (INR)'] < 0]
```

```python
df.loc[df['Discounted_Price (INR)'] <= 0] = np.nan
df['Discounted_Price (INR)'] = df['Discounted_Price (INR)'].fillna(df['Discounted_Price (INR)'].median())
```
